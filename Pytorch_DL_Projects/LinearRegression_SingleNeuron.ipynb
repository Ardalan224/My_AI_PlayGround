```python
import pandas as pd
import torch
from torch import nn
```

```python
# Reading the car prices csv data using Pandas from dataset directory
# License: CC BY 4.0 ATTRIBUTION 4.0 INTERNATIONAL
# => https://creativecommons.org/licenses/by/4.0/

# Source: TAEEF NAJIB,
# => https://www.kaggle.com/datasets/taeefnajib/used-car-price-prediction-dataset/data

df = pd.read_csv("path_to_file.csv")
```

```python
# Preparing the data as pandas dataframe
car_age = df["model_year"].max() - df["model_year"]

car_milage = df["milage"]
car_milage = car_milage.str.replace(",", "")
car_milage = car_milage.str.replace(" mi.", "")
car_milage = car_milage.astype(int)

car_price = df["price"]
car_price = car_price.str.replace("$", "")
car_price = car_price.str.replace(",", "")
car_price = car_price.astype(int)
```

```python
# Creating X and Y tensor data and normalizing them
X = torch.column_stack(
    [
        torch.tensor(car_age, dtype=torch.float32),
        torch.tensor(car_milage, dtype=torch.float32),
    ]
)
X_mean = X.mean(axis=0)
X_std = X.std(axis=0)
X = (X - X_mean) / X_std

Y = torch.tensor(car_price, dtype=torch.float32).reshape((-1, 1))
Y_mean = Y.mean()
Y_std = Y.std()
Y = (Y - Y_mean) / Y_std
```

```python
model = nn.Linear(2, 1)
loss_fn = torch.nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.001)

for i in range(0, 20000):
    # Training process
    optimizer.zero_grad()
    outputs = model(X)
    loss = loss_fn(outputs, Y)
    loss.backward()
    optimizer.step()

    if i % 100 == 0:
        print(loss)
```

```python
X_data = torch.tensor([[7, 15000], [3, 15000], [7, 30000]], dtype=torch.float32)

prediction = model((X_data - X_mean) / X_std)
print(prediction * Y_std + Y_mean)
```
